```python
import os
import re
import csv
import glob

# 批量修复所有历任特首原本下载的CSV和TXT文件名，统一格式为 DD-MM-YYYY
base_dir = "/content/drive/MyDrive/Study/HKU/POLI3148 Data Science in Politics and Public Administration/Final Project/Data/Data_Raw/Speeches_and_Articles"
# 若在本地Windows执行，请取消下面这行的注释并修改盘符
# base_dir = "g:/My Drive/Study/HKU/POLI3148 Data Science in Politics and Public Administration/Final Project/Data/Data_Raw/Speeches_and_Articles"

leaders = ["Carrie_Lam", "Leung_Chun-ying", "Donald_Tsang", "Tung_Chee-hwa", "John_Lee"]
langs = ["Traditional_Chinese", "Simplified_Chinese"]

def pad_date(date_str):
    # 匹配 YYYY-M-D 格式
    m = re.match(r'^(\d{4})[./\-](\d{1,2})[./\-](\d{1,2})$', date_str)
    if m:
        return f"{int(m.group(3)):02d}-{int(m.group(2)):02d}-{m.group(1)}"
    # 匹配 D-M-YYYY 格式
    m2 = re.match(r'^(\d{1,2})[./\-](\d{1,2})[./\-](\d{4})$', date_str)
    if m2:
        return f"{int(m2.group(1)):02d}-{int(m2.group(2)):02d}-{m2.group(3)}"
    return date_str

def fix_csv_and_files(folder):
    if not os.path.exists(folder):
        return
    csv_file = os.path.join(folder, "Press_Release_Index.csv")
    
    # 1. 重命名 txt 文件
    for filename in os.listdir(folder):
        if not filename.endswith(".txt"):
            continue
        # 提取文件名前面的日期部分 e.g., 1-2-2012_something.txt or 2017-06-30_something.txt
        m = re.match(r'^(\d{4}[./\-]\d{1,2}[./\-]\d{1,2}|\d{1,2}[./\-]\d{1,2}[./\-]\d{4})_(.+)$', filename)
        if m:
            old_date = m.group(1)
            rest = m.group(2)
            new_date = pad_date(old_date)
            if old_date != new_date:
                old_path = os.path.join(folder, filename)
                new_path = os.path.join(folder, f"{new_date}_{rest}")
                if not os.path.exists(new_path):
                    os.rename(old_path, new_path)
                    print(f"Renamed file: {filename} -> {new_date}_{rest}")
                else:
                    # 如果新名字已存在，说明原来下载重叠了
                    os.remove(old_path)

    # 2. 修复 CSV
    if os.path.exists(csv_file):
        changed = False
        with open(csv_file, 'r', encoding='utf-8-sig', newline='') as f:
            lines = f.readlines()
            
        if not lines:
            return
            
        new_lines = []
        for line in lines:
            if ';' in line:
                parts = line.split(';', 1)
                d = parts[0].strip()
                new_d = pad_date(d)
                if d != new_d:
                    changed = True
                new_lines.append(f"{new_d}; {parts[1].strip()}\n")
            else:
                new_lines.append(line)
                
        if changed:
            with open(csv_file, 'w', encoding='utf-8-sig', newline='') as f:
                f.writelines(new_lines)
            print(f"Fixed dates in CSV: {csv_file}")

# 执行批量修复
for leader in leaders:
    for lang in langs:
        fix_csv_and_files(os.path.join(base_dir, leader, lang))

print("🎉 全部本地文件名和CSV日期修正完毕！所有日期已统一强制为 DD-MM-YYYY。")
```

```python
import os
import requests
import time
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import re

# 设定李家超时期演辞/文章保存路径
base_dir_jl = os.path.join(
    project_path,
    "Data",
    "Data_Raw",
    "Speeches_and_Articles",
    "John_Lee",
)

save_dirs_jl = {
    "simplified": os.path.join(base_dir_jl, "Simplified_Chinese"),
    "traditional": os.path.join(base_dir_jl, "Traditional_Chinese"),
}

for d in save_dirs_jl.values():
    os.makedirs(d, exist_ok=True)

# 香港政府现任特首网站通过 XML 文件异步加载各个年份的数据
# 我们通过直接请求其背后的统一 XML 数据源，即可最安全、无遗漏地拿到所有演辞链接
jl_xml_urls = [
    "https://www.ceo.gov.hk/public/xml/media-elect.xml" # 候任期间
] + [f"https://www.ceo.gov.hk/public/xml/media-{year}.xml" for year in range(2022, 2028)]

def get_jl_article_links(url_list):
    # 此处收集繁体(tc)和简体(sc)的演辞链接
    article_links = {"traditional": [], "simplified": []}
    
    for url in url_list:
        try:
            response = requests.get(url, timeout=10)
            if response.status_code != 200:
                continue 
                
            print(f"正在分析 XML 数据源: {url}")
            response.encoding = 'utf-8' 
            soup = BeautifulSoup(response.text, 'html.parser')
            
            for year_tag in soup.find_all('year'):
                year = year_tag.get('value')
                if not year: continue
                
                for month_tag in year_tag.find_all('month'):
                    month = month_tag.get('value')
                    for day_tag in month_tag.find_all('day'):
                        day = day_tag.get('value')
                        
                        # 演辞类型为 type='speech'
                        for press in day_tag.find_all('press', type='speech'):
                            # 提取繁体 (tcTitle -> tctitle)
                            tc_title_tag = press.find('tctitle')
                            if tc_title_tag and tc_title_tag.get('url'):
                                title = tc_title_tag.get_text(separator=' ', strip=True)
                                href = tc_title_tag.get('url')
                                if title and href:
                                    date_text = f"{day}-{month}-{year}"
                                    if not href.startswith('http'):
                                        href = urljoin("https://www.ceo.gov.hk/", href)
                                    if not any(link == href for d, t, link in article_links["traditional"]):
                                        article_links["traditional"].append((date_text, title, href))
                                        
                            # 提取简体 (scTitle -> sctitle)
                            sc_title_tag = press.find('sctitle')
                            if sc_title_tag and sc_title_tag.get('url'):
                                title_sc = sc_title_tag.get_text(separator=' ', strip=True)
                                href_sc = sc_title_tag.get('url')
                                if title_sc and href_sc:
                                    date_text_sc = f"{day}-{month}-{year}"
                                    if not href_sc.startswith('http'):
                                        href_sc = urljoin("https://www.ceo.gov.hk/", href_sc)
                                    if not any(link == href_sc for d, t, link in article_links["simplified"]):
                                        article_links["simplified"].append((date_text_sc, title_sc, href_sc))
                                        
        except Exception as e:
            print(f"❌ 读取 XML 数据失败 {url}: {e}")
            
    return article_links

def scrape_and_save_jl_article(date_text, title, url, save_dir):
    try:
        safe_title = "".join([c for c in title if c not in '\\/:*?"<>|']).strip()
        if not safe_title:
            safe_title = url.split('/')[-1].replace('.html', '').replace('.htm', '')
            
        safe_date = date_text.replace('.', '-')
        base_name = f"{safe_date}_{safe_title}"
        target_path = os.path.join(save_dir, f"{base_name}.txt")
        
        # 如果文件已存在且内容正确则直接返回
        if os.path.exists(target_path):
            with open(target_path, 'r', encoding='utf-8') as f:
                content = f.read(500)
                if url in content: return True
                
        # 请求页面，引入 User-Agent 和 requests.Session 防反爬
        session = requests.Session()
        headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"}
        
        try:
            res = session.get(url, headers=headers, timeout=15)
        except requests.exceptions.TooManyRedirects:
            direct_url = url.replace("http://sc.info.gov.hk/gb/", "http://").replace("https://sc.info.gov.hk/gb/", "https://")
            res = session.get(direct_url, headers=headers, timeout=15)
            
        res.encoding = res.apparent_encoding if res.apparent_encoding else 'utf-8'
        soup = BeautifulSoup(res.text, 'html.parser')
        
        # 获取正文文本框
        content_div = soup.find('span', id='pressrelease') or \
                      soup.find('div', id='pressrelease') or \
                      soup.find('div', class_='content') or \
                      soup.body
            
        if content_div:
            for tag in content_div(['script', 'style', 'nav']): tag.decompose()
            text = content_div.get_text(separator='\n', strip=True)
            if not text: return False
            
            with open(target_path, 'w', encoding='utf-8') as f:
                f.write(f"标题: {title}\n链接: {url}\n" + "-" * 50 + "\n" + text)
            print(f"✅ 保存成功: {os.path.basename(target_path)}")
            return True
            
        return False
    except Exception as e:
        print(f"❌ 抓取失败: {url} - 错误: {e}")
        return False

# 1. 查找所有 XML 中的演辞链接
all_jl_links = get_jl_article_links(jl_xml_urls)
print(f"共提取到繁体演辞链接: {len(all_jl_links['traditional'])} 篇")
print(f"共提取到简体演辞链接: {len(all_jl_links['simplified'])} 篇")

# 2. 制作 CSV 目录
for lang_key, dir_path in save_dirs_jl.items():
    csv_file = os.path.join(dir_path, "Speeches_Index.csv")
    with open(csv_file, "w", encoding="utf-8-sig") as f:
        f.write("Date; Title; URL\n")
        for date_text, title, url in all_jl_links[lang_key]:
            f.write(f"{date_text}; {title}; {url}\n")
    print(f"✅ {lang_key}版 目录 CSV 表格已保存至: {csv_file}")

# 3. 下载流程，加入了失败重试机制
def download_articles(articles_list, save_dir, lang_name):
    print(f"\n开始抓取并保存李家超时期{lang_name}版演辞...")
    failed_links = []
    
    for i, (date_text, title, url) in enumerate(articles_list):
        if i > 0 and i % 50 == 0:
            print(f"{lang_name}进度: 已处理 {i}/{len(articles_list)} 篇文章...")
            
        success = scrape_and_save_jl_article(date_text, title, url, save_dir)
        if not success:
            failed_links.append((date_text, title, url))
        time.sleep(0.3)
        
    return failed_links

# 统一下载并记录失败项
failed_traditional = download_articles(all_jl_links['traditional'], save_dirs_jl['traditional'], "繁体")
failed_simplified = download_articles(all_jl_links['simplified'], save_dirs_jl['simplified'], "简体")

# 4. 遗漏/失败重试
def retry_failed(failed_list, save_dir, lang_name):
    if not failed_list:
        return
    print(f"\n--- 开始补全抓取 {lang_name} 遗漏/失败的演辞，共 {len(failed_list)} 篇 ---")
    still_failed = []
    for date_text, title, url in failed_list:
        print(f"重试: {url}")
        success = scrape_and_save_jl_article(date_text, title, url, save_dir)
        if not success:
            still_failed.append(url)
        time.sleep(1) # 重试时拉长间隔
        
    if still_failed:
        print(f"⚠️ {lang_name} 仍有 {len(still_failed)} 篇抓取失败，请检查网络或链接。")
    else:
        print(f"🎉 {lang_name} 失败项已全部补全！")

retry_failed(failed_traditional, save_dirs_jl['traditional'], "繁体")
retry_failed(failed_simplified, save_dirs_jl['simplified'], "简体")

# 进行数量核对
count_traditional = len([f for f in os.listdir(save_dirs_jl['traditional']) if f.endswith('.txt')])
count_simplified = len([f for f in os.listdir(save_dirs_jl['simplified']) if f.endswith('.txt')])
print("\n--- 任务完成 ---")
print(f"理论繁体演辞数: {len(all_jl_links['traditional'])} | 实际抓取数: {count_traditional}")
print(f"理论简体演辞数: {len(all_jl_links['simplified'])} | 实际抓取数: {count_simplified}")
if count_traditional == len(all_jl_links['traditional']) and count_simplified == len(all_jl_links['simplified']):
    print("✨ 所有任务圆满完成，数量无误！")
else:
    print("⚠️ 抓取数量与理论值有出入，请检查是否存在重复链接或异常页面。")
```

```python
import os
import csv
import requests
import time
from bs4 import BeautifulSoup
from urllib.parse import urljoin

base_dir_cl = os.path.join(
    project_path,
    "Data",
    "Data_Raw",
    "Speeches_and_Articles",
    "Carrie_Lam"
)

save_dirs_cl = {
    "simplified": os.path.join(base_dir_cl, "Simplified_Chinese"),
    "traditional": os.path.join(base_dir_cl, "Traditional_Chinese"),
}

session = requests.Session()
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,application/pdf,*/*;q=0.8"
}

def force_download_pdfs_from_csv(lang_name, lang_dir):
    csv_file = os.path.join(lang_dir, "Speeches_Index.csv")
    if not os.path.exists(csv_file):
        print(f"❌ 找不到对应索引文件，请确保之前跑过提取目录的代码: {csv_file}")
        return
        
    print(f"\n--- 开始直接下载 {lang_name} 缺失的 PDF ---")
    
    # 1. 提取 CSV 里的所有链接
    links = []
    with open(csv_file, "r", encoding="utf-8-sig") as f:
        reader = csv.reader(f, delimiter=';')
        next(reader, None) # 跳过表头
        for row in reader:
            if len(row) >= 3:
                links.append((row[0].strip(), row[1].strip(), row[2].strip()))
                
    missing_count = 0
    success_count = 0
    
    # 2. 遍历链接检查是否有缺失
    for date_text, title, url in links:
        safe_title = "".join([c for c in title if c not in '\\/:*?"<>|']).strip()
        if not safe_title:
            safe_title = url.split('/')[-1].replace('.html', '').replace('.htm', '')
        safe_date = date_text.replace('.', '-').replace('/', '-')
        base_name = f"{safe_date}_{safe_title}"
        
        target_txt = os.path.join(lang_dir, f"{base_name}.txt")
        target_pdf = os.path.join(lang_dir, f"{base_name}.pdf")
        
        # 判断：如果有正常的 txt，或者已经下载了 pdf，则跳过
        if os.path.exists(target_pdf) and os.path.getsize(target_pdf) > 1024:
            continue
        if os.path.exists(target_txt) and os.path.getsize(target_txt) > 200:
            continue
            
        missing_count += 1
        print(f"缺漏项目 -> 尝试强制下载 PDF: {title} ({url})")
        
        try:
            # 去除冗余的简繁体跳转网关，获取直连
            direct_url = url.replace("http://sc.info.gov.hk/gb/", "http://").replace("https://sc.info.gov.hk/gb/", "https://")
            res = session.get(direct_url, headers=headers, timeout=20, stream=True)
            
            content_type = res.headers.get('Content-Type', '').lower()
            
            # 情况1: 本身就是 PDF / 附件响应
            if 'application/pdf' in content_type or url.lower().endswith('.pdf'):
                with open(target_pdf, 'wb') as f:
                    for chunk in res.iter_content(chunk_size=8192):
                        if chunk: f.write(chunk)
                print(f"  ✅ 成功下载原生 PDF: {os.path.basename(target_pdf)}")
                success_count += 1
                continue
                
            # 情况2: 本身是 HTML 网页，但内容内嵌了 PDF 或包含了直接的 a 标签
            res.encoding = 'utf-8'
            soup = BeautifulSoup(res.text, 'html.parser')
            pdf_link = None
            
            # A: 查找嵌入式 iframe / embed
            embed = soup.find(['embed', 'iframe', 'object'])
            if embed and embed.get('src') and embed.get('src').lower().endswith('.pdf'):
                pdf_link = embed['src']
            else:
                # B: 查找纯 A 标签引用的 .pdf
                a_tag = soup.find('a', href=lambda x: x and x.lower().endswith('.pdf'))
                if a_tag:
                    pdf_link = a_tag['href']
                    
            if pdf_link:
                pdf_full_url = urljoin(direct_url, pdf_link)
                print(f"  🔍 从 HTML 网页中提取到隐藏 PDF 链接: {pdf_full_url}")
                pdf_res = session.get(pdf_full_url, headers=headers, timeout=20, stream=True)
                if pdf_res.status_code == 200:
                    with open(target_pdf, 'wb') as f:
                        for chunk in pdf_res.iter_content(chunk_size=8192):
                            if chunk: f.write(chunk)
                    print(f"  ✅ 成功提取并下载网页内嵌的 PDF")
                    success_count += 1
                else:
                    print(f"  ❌ 内嵌 PDF 下载失败，状态码: {pdf_res.status_code}")
            else:
                print(f"  ❌ 网页中没有找到 PDF 附件线索")
                
        except Exception as e:
            print(f"  ❌ 网络或存取错误: {e}")
            
        time.sleep(0.5) # 控制请求频率防封IP
        
    print(f"[{lang_name} 结束] 检测到缺漏: {missing_count} 篇，本次成功补全 PDF: {success_count} 篇！")

# 跳过网页爬取，直接通过已有的 CSV 索引目录强制下载遗漏的 PDF
force_download_pdfs_from_csv("繁体", save_dirs_cl["traditional"])
force_download_pdfs_from_csv("简体", save_dirs_cl["simplified"])
```

```python
import os
import requests
import time
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import re
import pandas as pd

# 设定董建华时期演辞(Speeches)保存路径
save_dir_ch = os.path.join(project_path, "Data", "Data_Raw", "Speeches_and_Articles", "Tung_Chee-hwa", "Traditional_Chinese")
os.makedirs(save_dir_ch, exist_ok=True)

# 董建华演辞主入口基准（修复访问限制，直接指定主索引文件）
ch_base_url = "https://www.ceo.gov.hk/archive/97-05/speech/cesp-c.htm"

def decode_hk_html(content):
    """
    终极乱码杀手解码器：强制优先尝试 UTF-8，再兜底 big5-hkscs。
    """
    encodings = ['utf-8', 'big5-hkscs', 'big5', 'gb18030', 'gbk']
    for enc in encodings:
        try:
            return content.decode(enc, errors='strict')
        except UnicodeDecodeError:
            continue
    return content.decode('big5-hkscs', errors='ignore')

def is_garbled(text):
    bad_chars = ['ä', '¤', '³', '¡', 'æ', 'å', 'ç', 'é', 'è', '½', '¿', 'œ', 'Ð', 'Â', 'Å', 'Ã', 'Î', 'ï']
    if any(c in text for c in bad_chars):
        return True
    ch_count = sum(1 for c in text if '\u4e00' <= c <= '\u9fa5')
    if len(text) > 50 and ch_count < len(text) * 0.1:
        return True
    return False

def clean_text_newlines(text):
    lines = [line.strip() for line in text.split('\n') if line.strip()]
    return '\n'.join(lines)

def guess_date_from_text(text, url):
    match = re.search(r'(?:(一九九[七八九]|二[〇零0O][〇零0O][〇一二三四五])年)?([一二三四五六七八九十廿卅]{1,3})月([一二三四五六七八九十廿卅]{1,3})日', text)
    if match:
        y_zh, m_zh, d_zh = match.groups()
        m = zh_num_map.get(m_zh)
        d = zh_num_map.get(d_zh)
        if m and d:
            year = ""
            if y_zh:
                year = zh_year_map.get(y_zh, "")
            if not year:
                y_match = re.search(r'cesp-c0(\d)', url)
                if y_match:
                    year = "200" + y_match.group(1)
                elif "97cesp" in url:
                    year = "1997"
                elif "cesp-c98" in url:
                    year = "1998"
                elif "cesp-c99" in url:
                    year = "1999"
                elif "cesp-c00" in url:
                    year = "2000"
                else:
                    y_match2 = re.search(r'(\d{4})', url)
                    if y_match2: year = y_match2.group(1)
            if year:
                return f"{year}-{m.zfill(2)}-{d.zfill(2)}"
                
    match2 = re.search(r'(\d{4})年(\d{1,2})月(\d{1,2})日', text)
    if match2: return f"{match2.group(1)}-{match2.group(2).zfill(2)}-{match2.group(3).zfill(2)}"
        
    match3 = re.search(r'(\d{1,2})[\.\-\/](\d{1,2})[\.\-\/](\d{4})', text)
    if match3: return f"{match3.group(3)}-{match3.group(2).zfill(2)}-{match3.group(1).zfill(2)}"
    
    return "未知日期"

def get_ch_monthly_pages(base_url):
    print(f"正在读取主页寻找所有年度归档页面: {base_url}")
    try:
        response = requests.get(base_url, timeout=10)
        html_text = decode_hk_html(response.content)
        soup = BeautifulSoup(html_text, 'html.parser')
        
        pages = set()
        pages.add(base_url)
        
        for option in soup.find_all('option'):
            if option.get('value') and '.htm' in option['value']:
                pages.add(urljoin(base_url, option['value']))
                    
        for a in soup.find_all('a', href=True):
            href = a['href']
            if re.search(r'(cesp|speech|index).*\.htm', href, re.IGNORECASE):
                full_url = urljoin(base_url, href)
                if 'ceo.gov.hk/archive/97-05/' in full_url and full_url not in pages:
                    pages.add(full_url)
                    
        return list(pages)
    except Exception as e:
        print(f"❌ 读取 {base_url} 失败: {e}")
        return [base_url]

def get_ch_article_links(url_list):
    article_links = []
    for url in url_list:
        print(f"正在分析该归档页寻找演辞: {url}")
        try:
            response = requests.get(url, timeout=10)
            html_text = decode_hk_html(response.content)
            soup = BeautifulSoup(html_text, 'html.parser')
            
            for a in soup.find_all('a', href=True):
                href = a['href']
                
                from bs4 import NavigableString
                title_parts = []
                for child in a.children:
                    if isinstance(child, NavigableString):
                        title_parts.append(child.strip())
                    elif child.name not in ['a', 'p', 'div', 'td', 'tr', 'table', 'li', 'ul']:
                        title_parts.append(child.get_text(strip=True))
                title = " ".join([p for p in title_parts if p]).strip()
                
                if not title or len(title) < 2 or title.lower() in ['back', 'top', 'home', '主页', '去', 'go', '主頁', '页首', '頁首', '回到页首', '回到頁首']:
                    continue
                # 拦截类似 "一月", "十一月" 这种纯月份导航标题
                if re.match(r'^[一二三四五六七八九十]{1,2}月$', title):
                    continue
                    
                href = re.sub(r'[\r\n\s]+', '', href)
                
                # 拦截网页内锚点跳转链接
                if href.startswith('#'):
                    continue
                
                # 排除根目录自身
                if re.search(r'(?:cesp-c\d*|97cesp-c|ceprig_c|index)\.htm$', href, re.IGNORECASE):
                    continue
                        
                date_text = "未知日期"
                if 'info.gov.hk/gia/general' in href:
                    url_date_match = re.search(r'/(\d{4})(\d{2})/(\d{2})/', href)
                    if url_date_match:
                        date_text = f"{url_date_match.group(1)}-{url_date_match.group(2)}-{url_date_match.group(3)}"
                
                if date_text == "未知日期":
                    parent_text = a.parent.get_text(separator=' ', strip=True)[:200]
                    match1 = re.search(r'(\d{1,2}[\.\-\/]\d{1,2}[\.\-\/]\d{4})', parent_text)
                    if match1:
                        date_text = match1.group(1).replace('/', '-').replace('.', '-')
                    else:
                        match2 = re.search(r'(\d{4})年(\d{1,2})月(\d{1,2})日', parent_text)
                        if match2:
                            date_text = f"{match2.group(1)}-{match2.group(2).zfill(2)}-{match2.group(3).zfill(2)}"
                
                if not href.startswith('http'):
                    if 'isd/news' in href or 'gia/general' in href:
                        match = re.search(r'(isd/news.*|gia/general.*)', href)
                        if match:
                            href = "http://www.info.gov.hk/" + match.group(1)
                    else:
                        href = urljoin(url, href)
                    
                if not any(link == href for d, t, link in article_links):
                    article_links.append((date_text, title, href))
        except Exception as e:
            print(f"❌ 读取该页 {url} 失败: {e}")
            
    return article_links

def scrape_and_save_ch_article(date_text, title, url):
    safe_title = "".join([c for c in title if c not in '\\/:*?"<>|']).strip()
    if len(safe_title) > 80: safe_title = safe_title[:80] + "..."
    if not safe_title: safe_title = url.split('/')[-1].replace('.html', '').replace('.htm', '')
        
    # 彻底封杀无实质内容的空壳目录（针对历史残留文件）
    if re.match(r'^([一二三四五六七八九十]{1,2}月|.*页首.*|.*頁首.*|.*主页.*|.*主頁.*)$', safe_title):
        # 顺手把本地已经下载的这些垃圾 txt 删掉
        for trash in os.listdir(save_dir_ch):
            if "未知日期" in trash and safe_title in trash:
                trash_path = os.path.join(save_dir_ch, trash)
                if os.path.exists(trash_path):
                    try:
                        os.remove(trash_path)
                        print(f"🗑️ 已删掉无用垃圾历史文件: {trash}")
                    except: pass
        return None

    final_date = date_text
    text_content = ""
    existing_file = None
    
    for filename in os.listdir(save_dir_ch):
        if filename.endswith('.txt') and safe_title in filename:
            path = os.path.join(save_dir_ch, filename)
            try:
                with open(path, 'r', encoding='utf-8') as f:
                    content_read = f.read(2000)
                    if url in content_read:
                        existing_file = path
                        break
            except Exception:
                pass
    
    needs_download = True
    # 判断是否可以复用本地文件
    if existing_file:
        with open(existing_file, 'r', encoding='utf-8') as f:
            text_content = f.read()
        if not is_garbled(text_content) and "對不起，我們找不到" not in text_content and "对不起，我们找不到" not in text_content and "Sorry, the page" not in text_content:
            needs_download = False
            
            # 如果是已经处理好的正确文件，直接计算 final_date，提前跳出以节省全方位解析时间
            cleaned_text = clean_text_newlines(text_content)
            if final_date == "未知日期" or final_date == "":
                guessed_date = guess_date_from_text(cleaned_text, url)
                if guessed_date != "未知日期":
                    final_date = guessed_date
            
            safe_date = final_date.replace('.', '-').replace('/', '-')
            base_name = f"{safe_date}_{safe_title}"
            expected_path = os.path.join(save_dir_ch, f"{base_name}.txt")
            
            # 如果文件名已经完好，并且内容没有冗余换行，直接快速跳过
            if expected_path == existing_file and text_content == cleaned_text:
                print(f"⏩ 完好无损，直接跳过: {os.path.basename(existing_file)}")
                return final_date
            
    for filename in os.listdir(save_dir_ch):
        if "未知日期" in filename and safe_title in filename:
            trash_path = os.path.join(save_dir_ch, filename)
            if trash_path != existing_file and os.path.exists(trash_path):
                try: os.remove(trash_path)
                except: pass
            
    if needs_download:
        try:
            res = requests.get(url, timeout=10)
            html_text = decode_hk_html(res.content)
            soup = BeautifulSoup(html_text, 'html.parser')
            
            content_div = soup.find('div', id='pressrelease') or soup.find('div', id='content') or soup.find('td', class_='content') or soup.body
            if content_div:
                for tag in content_div(['script', 'style', 'nav']): tag.decompose()
                raw_text = content_div.get_text(separator='\n', strip=True)
                if raw_text:
                    if not is_garbled(raw_text):
                        text_content = f"标题: {title}\n链接: {url}\n" + "-" * 50 + "\n" + raw_text
        except Exception as e:
            print(f"❌ 抓取失败: {url} - 错误: {e}")
            
    if not text_content or "對不起，我們找不到你要的網頁" in text_content or "Sorry, the page you requested cannot be found" in text_content:
        return None

    cleaned_text = clean_text_newlines(text_content)
    
    if final_date == "未知日期" or final_date == "":
        guessed_date = guess_date_from_text(cleaned_text, url)
        if guessed_date != "未知日期":
            final_date = guessed_date
    
    safe_date = final_date.replace('.', '-').replace('/', '-')
    base_name = f"{safe_date}_{safe_title}"
    target_path = os.path.join(save_dir_ch, f"{base_name}.txt")
    
    counter = 1
    while os.path.exists(target_path) and target_path != existing_file:
        with open(target_path, 'r', encoding='utf-8') as f:
            if url in f.read(500):
                break 
        counter += 1
        target_path = os.path.join(save_dir_ch, f"{base_name}_{counter}.txt")
        
    should_rewrite = False
    if not existing_file:
        should_rewrite = True
        print(f"✅ 下载成功: {os.path.basename(target_path)}")
    elif needs_download:
        should_rewrite = True
        if existing_file and os.path.exists(existing_file):
            try: os.remove(existing_file)
            except: pass
        print(f"🔄 下载重新覆盖: {os.path.basename(target_path)}")
    elif target_path != existing_file:
        should_rewrite = True
        if existing_file and os.path.exists(existing_file):
            try: os.remove(existing_file)
            except: pass
        print(f"🔄 日期修复并重命名: {os.path.basename(target_path)}")
    elif text_content != cleaned_text:
        should_rewrite = True
        print(f"🧹 清理多余空行: {os.path.basename(target_path)}")
    else:
        # 文件存在、内容一致、路径准确，直接跳过并节省硬盘 I/O
        print(f"⏩ 文件已存在且无误，跳过下载: {os.path.basename(existing_file)}")
        
    if should_rewrite:
        with open(target_path, 'w', encoding='utf-8') as f:
            f.write(cleaned_text)
            
    return final_date

# --- 主执行流程 ---
monthly_pages = [
    "https://www.ceo.gov.hk/archive/97-05/speech/csprig.htm", # 2005
    "https://www.ceo.gov.hk/archive/97-05/speech/cesp-c04.htm",
    "https://www.ceo.gov.hk/archive/97-05/speech/cesp-c03.htm",
    "https://www.ceo.gov.hk/archive/97-05/speech/cesp-c02.htm",
    "https://www.ceo.gov.hk/archive/97-05/speech/cesp-c01.htm",
    "https://www.ceo.gov.hk/archive/97-05/speech/cesp-c00.htm",
    "https://www.ceo.gov.hk/archive/97-05/speech/cesp-c99.htm",
    "https://www.ceo.gov.hk/archive/97-05/speech/cesp-c98.htm",
    "https://www.ceo.gov.hk/archive/97-05/speech/97cesp-c.htm"
]
print(f"共定位到 {len(monthly_pages)} 个董建华演辞的汇总归档页。")

all_ch_articles = get_ch_article_links(monthly_pages)
print(f"共提取到 {len(all_ch_articles)} 篇演辞链接候选。")

final_ch_articles = []
print("开始抓取、智能处理乱码、去除无用换行并检测未知日期...")
for i, (date_text, title, url) in enumerate(all_ch_articles):
    if i > 0 and i % 50 == 0:
        print(f"进度: 已处理 {i}/{len(all_ch_articles)} 篇文章...")
    
    final_date_res = scrape_and_save_ch_article(date_text, title, url)
    if final_date_res is not None:
        final_ch_articles.append((final_date_res, title, url))
    time.sleep(0.3)
    # 中文日期映射（兜底定义）
    if 'zh_num_map' not in globals():
        zh_num_map = {
            '一': '1', '二': '2', '三': '3', '四': '4', '五': '5', '六': '6', '七': '7', '八': '8', '九': '9', '十': '10',
            '十一': '11', '十二': '12', '十三': '13', '十四': '14', '十五': '15', '十六': '16', '十七': '17', '十八': '18', '十九': '19',
            '二十': '20', '二十一': '21', '二十二': '22', '二十三': '23', '二十四': '24', '二十五': '25', '二十六': '26', '二十七': '27',
            '二十八': '28', '二十九': '29', '三十': '30', '三十一': '31',
            '廿一': '21', '廿二': '22', '廿三': '23', '廿四': '24', '廿五': '25', '廿六': '26', '廿七': '27', '廿八': '28', '廿九': '29',
            '卅': '30', '卅一': '31'
        }

    if 'zh_year_map' not in globals():
        zh_year_map = {
            '一九九七': '1997', '一九九八': '1998', '一九九九': '1999',
            '二〇〇〇': '2000', '二零零零': '2000',
            '二〇〇一': '2001', '二零零一': '2001',
            '二〇〇二': '2002', '二零零二': '2002',
            '二〇〇三': '2003', '二零零三': '2003',
            '二〇〇四': '2004', '二零零四': '2004',
            '二〇〇五': '2005', '二零零五': '2005'
        }

    # 导出前清洗+去重
    cleaned = []
    seen = set()
    for d, t, u in final_ch_articles:
        d = re.sub(r'[\r\n\t]+', ' ', str(d)).strip()
        t = re.sub(r'[\r\n\t]+', ' ', str(t)).strip()
        u = str(u).strip()
        key = (d, t, u)
        if key not in seen:
            seen.add(key)
            cleaned.append(key)

    final_ch_articles = sorted(cleaned, key=lambda x: x[0], reverse=True)
    print(f"✅ 去重后保留 {len(final_ch_articles)} 篇。")
csv_file_path_ch = os.path.join(save_dir_ch, "Speeches_Index.csv")

# 转换格式并保存统一标准的 CSV
df_ch = pd.DataFrame(final_ch_articles, columns=['Date', 'Title', 'URL'])
df_ch['Date'] = df_ch['Date'].apply(lambda x: re.sub(r'[\r\n\t]+', ' ', str(x)).strip())
df_ch['Title'] = df_ch['Title'].apply(lambda x: re.sub(r'[\r\n\t]+', ' ', str(x)).strip())
df_ch.to_csv(csv_file_path_ch, index=False, encoding='utf-8-sig')

print(f"✅ 目录 CSV 表格已保存并更新至: {csv_file_path_ch}")
print("🎉 董建华各年度重要演辞爬取完毕！")
```

```python
import os
import re
import csv

# 董建华演辞日期双重修复：
# 1) 从正文中文日期提取，例如「一九九七年十二月二十日（星期六）」
# 2) 从 URL 提取，例如 https://www.info.gov.hk/gia/general/200503/04/03040265.htm -> 2005-03-04
# 规则：能从 URL 明确解析的，优先用 URL；否则用正文日期；两者冲突时记录到复核表。

save_dir_ch = os.path.join(project_path, "Data", "Data_Raw", "Speeches_and_Articles", "Tung_Chee-hwa", "Traditional_Chinese")
csv_file_path_ch = os.path.join(save_dir_ch, "Speeches_Index.csv")

cn_digit_map = {
    '零': 0, '〇': 0, '一': 1, '二': 2, '三': 3, '四': 4, '五': 5, '六': 6, '七': 7, '八': 8, '九': 9,
    '十': 10, '廿': 20, '卅': 30
}

zh_year_map = {
    '一九九七': '1997', '一九九八': '1998', '一九九九': '1999',
    '二零零零': '2000', '二〇〇〇': '2000',
    '二零零一': '2001', '二〇〇一': '2001',
    '二零零二': '2002', '二〇〇二': '2002',
    '二零零三': '2003', '二〇〇三': '2003',
    '二零零四': '2004', '二〇〇四': '2004',
    '二零零五': '2005', '二〇〇五': '2005',
}


def normalize_int(token):
    token = str(token).strip()
    if token.isdigit():
        return int(token)
    if token in cn_digit_map:
        return cn_digit_map[token]
    if token == '十一':
        return 11
    if token == '十二':
        return 12
    if token == '十三':
        return 13
    if token == '十四':
        return 14
    if token == '十五':
        return 15
    if token == '十六':
        return 16
    if token == '十七':
        return 17
    if token == '十八':
        return 18
    if token == '十九':
        return 19
    if token == '二十':
        return 20
    if token == '二十一':
        return 21
    if token == '二十二':
        return 22
    if token == '二十三':
        return 23
    if token == '二十四':
        return 24
    if token == '二十五':
        return 25
    if token == '二十六':
        return 26
    if token == '二十七':
        return 27
    if token == '二十八':
        return 28
    if token == '二十九':
        return 29
    if token == '三十':
        return 30
    if token == '三十一':
        return 31
    return None


def safe_filename(text):
    text = re.sub(r'[\r\n\t]+', ' ', str(text)).strip()
    text = ''.join(c for c in text if c not in '\\/:*?"<>|')
    text = re.sub(r'\s+', ' ', text)
    return text[:120].strip()


def date_in_range(date_text):
    if not date_text:
        return False
    m = re.match(r'^(\d{4})-(\d{2})-(\d{2})$', date_text)
    if not m:
        return False
    year = int(m.group(1))
    return 1997 <= year <= 2005


def extract_date_from_url(url):
    url = str(url).strip()
    if not url:
        return None

    # https://www.info.gov.hk/gia/general/200503/04/03040265.htm -> 2005-03-04
    m = re.search(r'/((?:19|20)\d{2})(\d{2})/(\d{2})/\d+\.htm', url, re.IGNORECASE)
    if m:
        return f"{m.group(1)}-{int(m.group(2)):02d}-{int(m.group(3)):02d}"

    # 199812/23/1223155.htm 之类
    m = re.search(r'/((?:19|20)\d{2})(\d{2})/(\d{2})/', url)
    if m:
        return f"{m.group(1)}-{int(m.group(2)):02d}-{int(m.group(3)):02d}"

    # 20050304 这种兜底
    m = re.search(r'((?:19|20)\d{2})(\d{2})(\d{2})', url)
    if m:
        return f"{m.group(1)}-{int(m.group(2)):02d}-{int(m.group(3)):02d}"

    return None


def extract_date_from_text(text):
    text = re.sub(r'[\r\n\t]+', ' ', str(text))
    text = re.sub(r'\s+', ' ', text).strip()

    # 一九九七年十二月二十日（星期六）
    m = re.search(r'((?:一|二|三|四|五|六|七|八|九|零|〇){4})年\s*([一二三四五六七八九十廿卅零〇0-9]{1,3})月\s*([一二三四五六七八九十廿卅零〇0-9]{1,3})日', text)
    if m:
        year = zh_year_map.get(m.group(1))
        month = normalize_int(m.group(2))
        day = normalize_int(m.group(3))
        if year and month and day:
            return f"{year}-{month:02d}-{day:02d}"

    # 1997年12月20日
    m = re.search(r'(\d{4})年\s*(\d{1,2})月\s*(\d{1,2})日', text)
    if m:
        return f"{m.group(1)}-{int(m.group(2)):02d}-{int(m.group(3)):02d}"

    # 12月20日（缺年份）
    m = re.search(r'([一二三四五六七八九十廿卅零〇0-9]{1,3})月\s*([一二三四五六七八九十廿卅零〇0-9]{1,3})日', text)
    if m:
        month = normalize_int(m.group(1))
        day = normalize_int(m.group(2))
        if month and day:
            return (month, day)

    # 20.12.1997 / 20-12-1997 / 20/12/1997
    m = re.search(r'(\d{1,2})[\.\-/](\d{1,2})[\.\-/](\d{4})', text)
    if m:
        return f"{m.group(3)}-{int(m.group(2)):02d}-{int(m.group(1)):02d}"

    # 1997.12.20 / 1997-12-20 / 1997/12/20
    m = re.search(r'(\d{4})[\.\-/](\d{1,2})[\.\-/](\d{1,2})', text)
    if m:
        return f"{m.group(1)}-{int(m.group(2)):02d}-{int(m.group(3)):02d}"

    return None


def resolve_date(text_date, url_date, file_name):
    if url_date and date_in_range(url_date):
        if isinstance(text_date, str) and text_date != url_date and date_in_range(text_date):
            return url_date, True
        return url_date, False

    if isinstance(text_date, str) and date_in_range(text_date):
        return text_date, False

    if isinstance(text_date, tuple):
        month, day = text_date
        year = None
        if url_date and date_in_range(url_date):
            year = url_date[:4]
        else:
            m = re.match(r'^(\d{4})', file_name)
            if m:
                year = m.group(1)
        if year:
            return f"{year}-{month:02d}-{day:02d}", False

    return None, False


def parse_record_from_file(file_path):
    with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
        content = f.read()

    url_match = re.search(r'^链接[:：]\s*(.+)$', content, re.MULTILINE)
    title_match = re.search(r'^标题[:：]\s*(.+)$', content, re.MULTILINE)
    url = url_match.group(1).strip() if url_match else ''
    title = title_match.group(1).strip() if title_match else os.path.splitext(os.path.basename(file_path))[0]

    text_date = extract_date_from_text(content)
    url_date = extract_date_from_url(url)
    final_date, conflict = resolve_date(text_date, url_date, os.path.basename(file_path))

    return {
        'file_path': file_path,
        'file_name': os.path.basename(file_path),
        'title': title,
        'url': url,
        'text_date': text_date,
        'url_date': url_date,
        'final_date': final_date,
        'conflict': conflict,
        'content': content,
    }


records = []
for name in os.listdir(save_dir_ch):
    if name.lower().endswith('.txt'):
        records.append(parse_record_from_file(os.path.join(save_dir_ch, name)))

fixed_count = 0
review_rows = []
for record in records:
    original_path = record['file_path']
    final_date = record['final_date']
    title = record['title']
    url = record['url']
    content = record['content']

    if not final_date or not date_in_range(final_date):
        review_rows.append({
            'file': record['file_name'],
            'title': title,
            'url': url,
            'text_date': str(record['text_date']),
            'url_date': str(record['url_date']),
            'final_date': str(final_date or '')
        })
        continue

    title_slug = safe_filename(title)
    new_name = f"{final_date}_{title_slug}.txt"
    new_path = os.path.join(save_dir_ch, new_name)

    if os.path.abspath(new_path) != os.path.abspath(original_path):
        if os.path.exists(new_path):
            os.remove(original_path)
        else:
            os.rename(original_path, new_path)
        fixed_count += 1
        target_path = new_path
    else:
        target_path = original_path

    # 规范化文件内容头部，确保 URL 和标题仍然保留
    try:
        with open(target_path, 'w', encoding='utf-8') as f:
            f.write(content)
    except Exception:
        pass

# 用最终可验证日期重建 CSV
csv_rows = []
for record in records:
    if record['final_date'] and date_in_range(record['final_date']):
        csv_rows.append((record['final_date'], record['title'], record['url']))

csv_rows.sort(key=lambda row: row[0], reverse=True)
with open(csv_file_path_ch, 'w', encoding='utf-8-sig', newline='') as f:
    writer = csv.writer(f, delimiter=';')
    writer.writerow(['Date', 'Title', 'URL'])
    for row in csv_rows:
        writer.writerow(row)

print(f"✅ 董建华日期修复完成：共检查 {len(records)} 个 txt，重命名/修正 {fixed_count} 个。")
print(f"✅ 已重建 CSV: {csv_file_path_ch}")
if review_rows:
    review_path = os.path.join(save_dir_ch, 'date_review_needed.csv')
    with open(review_path, 'w', encoding='utf-8-sig', newline='') as f:
        writer = csv.DictWriter(f, fieldnames=['file', 'title', 'url', 'text_date', 'url_date', 'final_date'])
        writer.writeheader()
        writer.writerows(review_rows)
    print(f"⚠️ 仍有 {len(review_rows)} 条需要人工复核，已输出到: {review_path}")
else:
    print("✅ 没有发现需要额外人工复核的日期项。")
```
