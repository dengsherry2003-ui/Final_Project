```python
#董建华press release txt files
txt_dir = os.path.join(
    project_path, "Data", "Data_Raw", "Speeches_and_Articles", "Tung_Chee-hwa", "Traditional_Chinese"
)

tung_chee_hwa_txt = {}

if os.path.exists(txt_dir):
    tung_txt_files = sorted([f for f in os.listdir(txt_dir) if f.lower().endswith(".txt")])
    print(f"Found {len(tung_txt_files)} txt files in:\n{txt_dir}\n")

    for fname in tung_txt_files:
        file_path = os.path.join(txt_dir, fname)
        with open(file_path, "r", encoding="utf-8") as f:
            tung_chee_hwa_txt[fname] = f.read()

    print(
        "Loaded files:",
        list(tung_chee_hwa_txt.keys())[:10],
        "..." if len(tung_chee_hwa_txt) > 10 else ""
    )
else:
    print("Path not found:", txt_dir)
```

```python
# Load press release txt files for OTHER Chief Executives (excluding John Lee and Tung Chee-hwa)
import pandas as pd

press_release_root = os.path.join(project_path, "Data", "Data_Raw", "Speeches_and_Articles")
excluded_folders = {"John_Lee", "Tung_Chee-hwa"}

other_ce_txt = {}      # {folder_name: {filename: text}}
other_ce_meta = []     # summary table-like list

if os.path.exists(press_release_root):
    ce_folders = sorted([
        d for d in os.listdir(press_release_root)
        if os.path.isdir(os.path.join(press_release_root, d)) and d not in excluded_folders
    ])

    print(f"Found {len(ce_folders)} other CE folder(s): {ce_folders}\n")

    for ce_folder in ce_folders:
        trad_dir = os.path.join(press_release_root, ce_folder, "Traditional_Chinese")
        simp_dir = os.path.join(press_release_root, ce_folder, "Simplified_Chinese")

        source_dir = trad_dir if os.path.exists(trad_dir) else simp_dir
        ce_texts = {}

        if os.path.exists(source_dir):
            txt_files = sorted([f for f in os.listdir(source_dir) if f.lower().endswith(".txt")])
            for fname in txt_files:
                fpath = os.path.join(source_dir, fname)
                with open(fpath, "r", encoding="utf-8") as f:
                    ce_texts[fname] = f.read()

            other_ce_txt[ce_folder] = ce_texts
            other_ce_meta.append({
                "CE_Folder": ce_folder,
                "Source": "Traditional_Chinese" if source_dir == trad_dir else "Simplified_Chinese",
                "Files_Loaded": len(ce_texts)
            })

            print(f"[OK] {ce_folder}: loaded {len(ce_texts)} file(s) from {source_dir}")
        else:
            print(f"[SKIP] {ce_folder}: no Traditional_Chinese or Simplified_Chinese folder found")
else:
    print("Path not found:", press_release_root)

# Optional: quick summary
if other_ce_meta:
    summary_other_ce = pd.DataFrame(other_ce_meta)
    display(summary_other_ce)
```

# 4. Data Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt

# --- 1. 确保核心函数和映射存在 ---
ce_name_map = {
    "John_Lee": "John Lee",
    "Tung_Chee-hwa": "Tung Chee-hwa",
    "Carrie_Lam": "Carrie Lam",
    "Donald_Tsang": "Donald Tsang",
    "Leung_Chun-ying": "CY Leung"
}

def calculate_metrics(text_dict, keywords):
    """统一的统计计算逻辑"""
    if not text_dict:
        return 0, []
    full_text = "".join(str(v) for v in text_dict.values())
    total_chars = len(full_text)
    if total_chars == 0:
        return 0, []
    
    # 使用 keywords 统计词频
    counts = {kw: full_text.count(kw) for kw in keywords}
    total_hits = sum(counts.values())
    density = (total_hits / total_chars) * 10000
    top_5 = sorted(counts.items(), key=lambda x: x[1], reverse=True)[:5]
    return density, top_5

# --- 2. 收集数据 (重新计算，避免 NameError) ---
all_ce_data = []

# 列表化需要处理的对象
# 格式: (显示名称, 文本数据源)
targets = [
    ("John Lee", john_lee_txt),
    ("Tung Chee-hwa", tung_chee_hwa_txt)
]

# 将 targets 中的数据计算后加入列表
for display_name, txt_source in targets:
    if display_name in globals() or 'john_lee_txt' in locals() or 'john_lee_txt' in globals(): # 检查变量是否存在
        density, top5 = calculate_metrics(txt_source, mainlandization_keywords)
        all_ce_data.append({"Name": display_name, "Density": density, "Top5": top5})

# 处理 other_ce_txt 中的其他特首
if 'other_ce_txt' in globals():
    for ce_folder, ce_texts in other_ce_txt.items():
        density, top5 = calculate_metrics(ce_texts, mainlandization_keywords)
        readable_name = ce_name_map.get(ce_folder, ce_folder)
        all_ce_data.append({"Name": readable_name, "Density": density, "Top5": top5})

# --- 3. 检查数据是否收集成功 ---
if not all_ce_data:
    print("错误：未找到任何文本数据，请确保 john_lee_txt 等变量已正确加载。")
else:
    # 转为 DataFrame 并排序
    plot_df = pd.DataFrame(all_ce_data).sort_values(by="Density", ascending=False)

    # --- 4. 可视化 ---
    plt.figure(figsize=(12, 6))
    # 使用更现代的调色盘
    colors = plt.cm.viridis(pd.Series(range(len(plot_df))) / len(plot_df))

    bars = plt.bar(plot_df["Name"], plot_df["Density"], color=colors, alpha=0.8)

    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 0.1,
                 f'{height:.1f}', ha='center', va='bottom', fontweight='bold', fontsize=11)

    plt.title("Linguistic Evolution: 'Officialese' Density Across All CEs", fontsize=15, pad=20)
    plt.ylabel("Keyword Hits per 10,000 Characters", fontsize=12)
    plt.xticks(rotation=45)
    plt.grid(axis="y", linestyle="--", alpha=0.3)
    plt.tight_layout()
    plt.show()
    # --- Top 20 frequent keywords (overall) ---
    source_map = {
        "John Lee": john_lee_txt,
        "Tung Chee-hwa": tung_chee_hwa_txt
    }
    if 'other_ce_txt' in globals():
        for ce_folder, ce_texts in other_ce_txt.items():
            source_map[ce_name_map.get(ce_folder, ce_folder)] = ce_texts

    overall_counts = {kw: 0 for kw in mainlandization_keywords}
    for texts in source_map.values():
        full_text = "".join(str(v) for v in texts.values())
        for kw in overall_counts:
            overall_counts[kw] += full_text.count(kw)

    top20_overall = sorted(overall_counts.items(), key=lambda x: x[1], reverse=True)[:20]

    print("\nTop 20 frequent keywords (overall):")
    for i, (kw, cnt) in enumerate(top20_overall, 1):
        print(f"{i:>2}. {kw}: {cnt}")
        
    # --- 5. 打印对比详情 ---
    ce_top20_map = {}
    for ce_name, texts in source_map.items():
        full_text = "".join(str(v) for v in texts.values()) if isinstance(texts, dict) else ""
        counts = {kw: full_text.count(kw) for kw in mainlandization_keywords}
        ce_top20_map[ce_name] = sorted(counts.items(), key=lambda x: x[1], reverse=True)[:20]

    # 写回到 plot_df，便于后续查看/导出
    plot_df["Top20"] = plot_df["Name"].map(ce_top20_map)
    print("\n每位特首提及最多的 10 个关键词：")
    for ce_name in plot_df["Name"]:
        top20 = ce_top20_map.get(ce_name, [])
        print(f"\n{ce_name}:")
        non_zero_items = "、".join(f"{kw}({cnt})" for kw, cnt in top20 if cnt > 0)
        print(non_zero_items if non_zero_items else "（无匹配关键词）")

    print("\n" + "=" * 50)
    print(f"{'Chief Executive':<18} | {'Top 10 Frequent Keywords'}")
    print("-" * 50)
    for ce_name in plot_df["Name"]:
        top10 = ce_top20_map.get(ce_name, [])[:10]
        print(f"{ce_name:<18} | {top10}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# ==========================================
# 按年份统计官腔密度：柱状图（总体）+ 折线图（各特首）
# ==========================================

# 1. 函数：从文件名提取年份 (提取逻辑视文件名格式而定)
def extract_year(filename):
    # 假设文件名中含有类似 '1997', '2015' 的 4 位数字
    import re
    match = re.search(r'((?:19|20)\d{2})', filename)
    if match:
        return match.group(1)
    return None

# 2. 收集所有数据到按年份聚合的结构中
yearly_data = []

# 定义我们要处理的特首数据源
# 假设 john_lee_txt, tung_chee_hwa_txt 等变量已经由前面的单元格加载
all_sources = [
    ("John Lee", globals().get('john_lee_txt', {})),
    ("Tung Chee-hwa", globals().get('tung_chee_hwa_txt', {}))
]

# 如果加载了前面单元格写的 other_ce_txt
if 'other_ce_txt' in globals():
    for ce_folder, texts in other_ce_txt.items():
        # 转为英文名用于图表图例
        ce_name = ce_name_map.get(ce_folder, ce_folder)
        all_sources.append((ce_name, texts))

for ce_name, texts in all_sources:
    if not texts:
        continue
        
    for fname, content in texts.items():
        year = extract_year(fname)
        if not year:
            continue
            
        char_count = len(content)
        # 计算该篇文本包含的官话关键词命中总数
        hits = sum(content.count(kw) for kw in mainlandization_keywords)
        
        yearly_data.append({
            "CE": ce_name,
            "Year": int(year),
            "Hits": hits,
            "Chars": char_count
        })

df_raw = pd.DataFrame(yearly_data)

# 如果没有任何带有正确年份的数据，退出执行
if df_raw.empty:
    print("未能在文件名中提取到可用的年份信息，无法绘制图表。请检查命名格式。")
else:
    # 剔除可能明显不是目标年份的异常数据（比如提取到了网页里的杂乱数字）
    df_raw = df_raw[(df_raw['Year'] >= 1997) & (df_raw['Year'] <= 2026)]

    # 3. 聚合数据 
    # a. 按年份计算总体密度（用于柱状图）
    df_yearly = df_raw.groupby('Year').agg({'Hits': 'sum', 'Chars': 'sum'}).reset_index()
    df_yearly['Density'] = (df_yearly['Hits'] / df_yearly['Chars']) * 10000

    # b. 按 年份+特首 计算密度（用于折线图）
    df_ce_yearly = df_raw.groupby(['Year', 'CE']).agg({'Hits': 'sum', 'Chars': 'sum'}).reset_index()
    df_ce_yearly['Density'] = (df_ce_yearly['Hits'] / df_ce_yearly['Chars']) * 10000

    # ==========================================
    # 4. 绘制双轴图（柱状图 + 折线图）
    # ==========================================
    fig, ax1 = plt.subplots(figsize=(15, 7))

    # 调色盘
    ce_colors = {
        "Tung Chee-hwa": "#3498DB",  # 蓝
        "Donald Tsang": "#2ECC71",    # 绿
        "CY Leung": "#F1C40F",        # 黄
        "Carrie Lam": "#E67E22",      # 橙
        "John Lee": "#E74C3C",        # 红
    }

    # 绘制背景柱状图：展示该年总体的平均官腔密度
    ax1.bar(df_yearly['Year'], df_yearly['Density'], color='#95A5A6', alpha=0.3, label='Overall Average Density', zorder=1)

    # 绘制前景折线图：展示不同特首在该年的表现
    for ce_name in df_ce_yearly['CE'].unique():
        df_sub = df_ce_yearly[df_ce_yearly['CE'] == ce_name].sort_values('Year')
        # 确保该特首该年不是因为只有一两篇文章而导致的极端异常值（可选：Chars > 1000）
        df_sub = df_sub[df_sub['Chars'] > 2000]
        
        color = ce_colors.get(ce_name, "#9B59B6") # 默认给个紫色
        ax1.plot(df_sub['Year'], df_sub['Density'], marker='o', markersize=6, linewidth=3, 
                 color=color, label=ce_name, zorder=5)

    # 将 X 轴显式设置为整数年份
    min_year = df_yearly['Year'].min()
    max_year = df_yearly['Year'].max()
    ax1.set_xticks(np.arange(min_year, max_year + 1))
    ax1.set_xticklabels(np.arange(min_year, max_year + 1), rotation=45)

    # 设置标签与标题
    ax1.set_xlabel('Year', fontsize=12)
    ax1.set_ylabel("Keyword Hits per 10k Characters", fontsize=12)
    plt.title("Trend of Loyalty Signalling in CE Speeches by Year (1997-Present)", fontsize=16, pad=15)
    
    # 设定图例：合并 ax1 的内容
    handles, labels = ax1.get_legend_handles_labels()
    fig.legend(handles, labels, loc='upper left', bbox_to_anchor=(0.28, 0.88), frameon=True, fontsize=10)

    # 网格线
    ax1.grid(axis="y", linestyle="--", alpha=0.5)

    plt.tight_layout()
    plt.show()
```

```python
#差异显著性分析
import math

# Use existing aggregated stats
c_j, n_j = john_stats["total_hits"], john_stats["total_chars"]
c_t, n_t = tung_stats["total_hits"], tung_stats["total_chars"]

rate_j = c_j / n_j
rate_t = c_t / n_t
rr = rate_j / rate_t  # rate ratio (John Lee vs Tung Chee-hwa)

# Poisson rate ratio z-test (approx.)
se_log_rr = math.sqrt((1 / c_j) + (1 / c_t))
z = math.log(rr) / se_log_rr
p_value_two_sided = math.erfc(abs(z) / math.sqrt(2))

# 95% CI for rate ratio
ci_low = math.exp(math.log(rr) - 1.96 * se_log_rr)
ci_high = math.exp(math.log(rr) + 1.96 * se_log_rr)

print(f"John Lee rate: {rate_j*10000:.3f} per 10k chars")
print(f"Tung Chee-hwa rate: {rate_t*10000:.3f} per 10k chars")
print(f"Rate ratio (John/Tung): {rr:.3f}")
print(f"95% CI: [{ci_low:.3f}, {ci_high:.3f}]")
print(f"z = {z:.2f}, p-value = {p_value_two_sided:.3e}")

if p_value_two_sided < 0.05:
	print("Conclusion: The difference is statistically significant.")
else:
	print("Conclusion: The difference is NOT statistically significant.")
```

```python
import os
import re
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# ==================================================
# Year-by-year loyalty trend by Chief Executive
# ==================================================
# This cell rebuilds the six loyalty categories from the keyword cell in this notebook,
# counts keyword hits in each press-release text, and plots year-by-year densities.

notebook_candidates = []
if "project_path" in globals():
    notebook_candidates.append(Path(project_path) / "Code" / "CE_Speech_analysis.ipynb")
notebook_candidates.append(Path(r"g:\My Drive\Study\HKU\POLI3148 Data Science in Politics and Public Administration\Final Project\Code\CE_Speech_analysis.ipynb"))
notebook_candidates.append(Path.cwd() / "CE_Speech_analysis.ipynb")
notebook_candidates.append(Path.cwd() / "Code" / "CE_Speech_analysis.ipynb")

notebook_path = next((p for p in notebook_candidates if p.exists()), None)
if notebook_path is None:
    raise FileNotFoundError("Could not locate CE_Speech_analysis.ipynb for keyword parsing.")

# Prefer the canonical CE names in the chart.
ce_to_folder = {
    "Tung Chee-hwa": "Tung_Chee-hwa",
    "Donald Tsang": "Donald_Tsang",
    "Leung Chun-ying": "Leung_Chun-ying",
    "Carrie Lam": "Carrie_Lam",
    "John Lee": "John_Lee",
}

ce_label_map = {
    "Tung_Chee-hwa": "Tung Chee-hwa",
    "Donald_Tsang": "Donald Tsang",
    "Leung_Chun-ying": "Leung Chun-ying",
    "Carrie_Lam": "Carrie Lam",
    "John_Lee": "John Lee",
}

# Try to reuse the notebook's existing keyword cell by parsing the code source from disk.
def load_loyalty_categories_from_notebook(nb_path):
    with open(nb_path, "r", encoding="utf-8") as f:
        nb = json.load(f)

    target_source = None
    for cell in nb.get("cells", []):
        source = "".join(cell.get("source", []))
        if "personal_loyalty" in source and "Rhetorical_loyalty" in source:
            target_source = cell.get("source", [])
            break

    if target_source is None:
        raise ValueError("Could not find the loyalty keyword cell in the notebook.")

    category_aliases = {
        "personal_loyalty": "Personal loyalty",
        "Institutional_loyalty": "Institutional loyalty",
        "policy_loyalty": "Policy loyalty",
        "nationalistic_loyalty": "Nationalistic loyalty",
        "Ideological_loyalty": "Ideological loyalty",
        "Rhetorical_loyalty": "Rhetorical loyalty",
    }

    categories = {label: [] for label in category_aliases.values()}
    current = None
    for raw_line in target_source:
        line = raw_line.strip()
        if line.startswith('#"') and line.endswith('":'):
            key = line[2:-2]
            current = category_aliases.get(key)
            continue
        if current is None:
            continue
        matches = re.findall(r'"([^"]+)"|\'([^\']+)\'', line)
        for match in matches:
            token = match[0] or match[1]
            token = token.strip()
            if token:
                categories[current].append(token)

    # Remove duplicates while preserving order.
    for key, values in categories.items():
        seen = set()
        unique_values = []
        for value in values:
            if value not in seen:
                seen.add(value)
                unique_values.append(value)
        categories[key] = unique_values

    return categories


loyalty_categories = load_loyalty_categories_from_notebook(notebook_path)

# Fallback to the current kernel's keyword set if parsing fails for any reason.
if not loyalty_categories or not any(loyalty_categories.values()):
    if "mainlandization_keywords" in globals() and isinstance(mainlandization_keywords, (set, list, tuple)):
        fallback_keywords = list(mainlandization_keywords)
        loyalty_categories = {"Overall loyalty terms": fallback_keywords}
    else:
        raise ValueError("No loyalty keywords available.")

# Find the raw press-release directories.
base_press_release = Path(project_path) / "Data" / "Data_Raw" / "Speeches_and_Articles"
if not base_press_release.exists():
    base_press_release = Path(r"g:\My Drive\Study\HKU\POLI3148 Data Science in Politics and Public Administration\Final Project\Data\Data_Raw\Speeches_and_Articles")

ce_folders = [folder for folder in ce_to_folder.values() if (base_press_release / folder).exists()]


def extract_year_from_filename(file_name):
    name = str(file_name)
    patterns = [
        r'(19\d{2}|20\d{2})[-_](\d{2})[-_](\d{2})',
        r'(\d{2})[-_](\d{2})[-_](19\d{2}|20\d{2})',
        r'(19\d{2}|20\d{2})',
    ]
    for pattern in patterns:
        match = re.search(pattern, name)
        if match:
            groups = match.groups()
            if len(groups) == 3:
                if groups[0].isdigit() and len(groups[0]) == 4:
                    return int(groups[0])
                if groups[2].isdigit() and len(groups[2]) == 4:
                    return int(groups[2])
            if len(groups) == 1:
                return int(groups[0])
    return None


def count_keyword_hits(text, keywords):
    total = 0
    for kw in keywords:
        if not kw:
            continue
        total += len(re.findall(re.escape(kw), text))
    return total

records = []
for folder_name in ce_folders:
    folder_path = base_press_release / folder_name
    for lang_name in ["Traditional_Chinese", "Simplified_Chinese"]:
        lang_path = folder_path / lang_name
        if not lang_path.exists():
            continue
        for file_path in lang_path.glob("*.txt"):
            try:
                content = file_path.read_text(encoding="utf-8", errors="ignore")
            except Exception:
                continue
            year = extract_year_from_filename(file_path.name)
            if year is None:
                header_match = re.search(r"^(?:Date|日期)[:：]\s*(.+)$", content, flags=re.MULTILINE)
                if header_match:
                    year_match = re.search(r'(19\d{2}|20\d{2})', header_match.group(1))
                    if year_match:
                        year = int(year_match.group(1))
            if year is None:
                continue
            ce_label = ce_label_map.get(folder_name, folder_name)
            char_count = len(content)
            for category_name, keyword_list in loyalty_categories.items():
                hits = count_keyword_hits(content, keyword_list)
                density = (hits / char_count * 10000) if char_count else 0
                records.append({
                    "CE": ce_label,
                    "Year": year,
                    "Category": category_name,
                    "Hits": hits,
                    "Chars": char_count,
                    "Density": density,
                    "Language": lang_name,
                })

trend_df = pd.DataFrame(records)
if trend_df.empty:
    raise ValueError("No press-release records were collected for the loyalty trend chart.")

# Combine the two language copies by averaging the densities across the same year/CE/category.
trend_df = (
    trend_df.groupby(["CE", "Year", "Category"], as_index=False)
    .agg({"Hits": "sum", "Chars": "sum", "Density": "mean"})
)

category_order = [
    "Personal loyalty",
    "Institutional loyalty",
    "Policy loyalty",
    "Nationalistic loyalty",
    "Ideological loyalty",
    "Rhetorical loyalty",
]
existing_categories = [c for c in category_order if c in trend_df["Category"].unique()]
if not existing_categories:
    existing_categories = sorted(trend_df["Category"].unique())

ce_order = [ce for ce in ["Tung Chee-hwa", "Donald Tsang", "Leung Chun-ying", "Carrie Lam", "John Lee"] if ce in trend_df["CE"].unique()]

sns.set_theme(style="whitegrid", rc={"axes.edgecolor": "0.8", "grid.color": "0.9"})
# Use sharey=True to unify the y-axis across all subplots
fig, axes = plt.subplots(2, 3, figsize=(22, 12), sharex=True, sharey=True)
axes = axes.flatten()

# A more aesthetic palette
palette = {
    "Tung Chee-hwa": "#3498DB",   # blue
    "Donald Tsang": "#2ECC71",    # green
    "Leung Chun-ying": "#F1C40F", # yellow 
    "Carrie Lam": "#E67E22",      # orange
    "John Lee": "#E74C3C",        # red
}

for idx, category_name in enumerate(existing_categories):
    ax = axes[idx]
    subset = trend_df[trend_df["Category"] == category_name].copy()
    
    for ce_name in ce_order:
        ce_data = subset[subset["CE"] == ce_name].sort_values("Year")
        if ce_data.empty:
            continue
            
        color = palette.get(ce_name, "#333333")
        
        # Plot lines with a slight shadow/glow effect
        ax.plot(
            ce_data["Year"],
            ce_data["Density"],
            marker="o",
            linewidth=2.5,
            markersize=6,
            markeredgecolor="white",
            markeredgewidth=1.5,
            label=ce_name,
            color=color,
            alpha=0.9
        )
        
        # Optionally fill under the line to make it pop slightly
        ax.fill_between(
            ce_data["Year"], 
            0, 
            ce_data["Density"], 
            color=color, 
            alpha=0.05
        )
        
    ax.set_title(category_name, fontsize=15, fontweight="bold", pad=12)
    ax.set_xlabel("Year", fontsize=12)
    ax.set_ylabel("Density (Hits / 10k chars)", fontsize=12)
    
    all_years = sorted(trend_df["Year"].unique())
    # Display every 2 years to avoid crowding
    ax.set_xticks(np.arange(min(all_years), max(all_years)+1, 2))
    ax.tick_params(axis='x', rotation=45, labelsize=11)
    ax.tick_params(axis='y', labelsize=11)
    ax.grid(True, linestyle='--', alpha=0.6)
    
    # Remove top and right spines
    sns.despine(ax=ax, top=True, right=True)

# Remove any unused axes
for j in range(len(existing_categories), len(axes)):
    fig.delaxes(axes[j])

# Common legend
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles, labels, 
    loc="upper center", 
    ncol=min(len(ce_order), 5), 
    frameon=True, 
    framealpha=0.9,
    edgecolor="lightgray",
    fontsize=13,
    bbox_to_anchor=(0.5, 0.94)
)

fig.suptitle("Evolution of Loyalty Signalling Density in CE Speeches by Category", fontsize=22, fontweight="bold", y=0.99)
fig.tight_layout(rect=[0, 0, 1, 0.9])
plt.show()

print("✅ 已生成优化后的 year-by-year loyalty 对比图。")
```
