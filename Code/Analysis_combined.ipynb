```python
import scipy.stats as stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# ==========================================
# 1. 数据准备：计算单篇文档的官腔密度（仅 speeches）
# ==========================================
ce_name_map = {
    "Donald_Tsang": "Donald Tsang",
    "Leung_Chun-ying": "CY Leung",
    "Carrie_Lam": "Carrie Lam",
    "John_Lee": "John Lee",
    "Tung_Chee-hwa": "Tung Chee-hwa",
}

records = []
for fname, content in globals().get("john_lee_speeches_txt", {}).items():
    records.append({
        "CE": "John Lee",
        "Hits": sum(content.count(kw) for kw in loyalty_signaling_keywords),
        "Chars": len(content),
    })
for fname, content in globals().get("tung_chee_hwa_speeches_txt", {}).items():
    records.append({
        "CE": "Tung Chee-hwa",
        "Hits": sum(content.count(kw) for kw in loyalty_signaling_keywords),
        "Chars": len(content),
    })
for ce_folder, texts in globals().get("other_ce_speeches_txt", {}).items():
    ce_label = ce_name_map.get(ce_folder, ce_folder)
    for fname, content in texts.items():
        records.append({
            "CE": ce_label,
            "Hits": sum(content.count(kw) for kw in loyalty_signaling_keywords),
            "Chars": len(content),
        })

df_raw = pd.DataFrame(records)
if df_raw.empty:
    raise ValueError("No speech documents loaded. Please run the speeches data loading cells first.")

# 在进行统计检验前，我们必须基于“每一篇演讲稿”计算密度，而不是年度总和
df_raw['Density'] = (df_raw['Hits'] / df_raw['Chars']) * 10000

# 剔除可能存在的缺失值或零字数异常文档
df_stats = df_raw[df_raw['Chars'] > 0].dropna(subset=['Density', 'CE'])

# ==========================================
# 2. 描述性统计与变异系数 (CV) 表格
# ==========================================
print("--- 历任特首用词频率统计汇总表 ---")
summary_stats = df_stats.groupby('CE')['Density'].agg(
    Count='count',
    Mean='mean',
    Median='median',
    Std='std'
).reset_index()

# 计算变异系数 CV (标准差 / 均值)
summary_stats['CV'] = summary_stats['Std'] / summary_stats['Mean']

# 按照时间顺序排序（假设你希望按照就任顺序查看）
# 这里我们自定义一个排序字典
ce_order = {"Tung Chee-hwa": 1, "Donald Tsang": 2, "CY Leung": 3, "Carrie Lam": 4, "John Lee": 5}
summary_stats['Order'] = summary_stats['CE'].map(ce_order)
summary_stats = summary_stats.sort_values('Order').drop('Order', axis=1).round(2)

print(summary_stats.to_string(index=False))
print("\n")

# ==========================================
# 3. 差异分析：单因素方差分析 (ANOVA)
# ==========================================
# 提取各特首的密度数据组
groups = [group['Density'].values for name, group in df_stats.groupby('CE')]

f_stat, p_value = stats.f_oneway(*groups)
print(f"--- 单因素方差分析 (ANOVA) ---")
print(f"F-statistic: {f_stat:.4f}")
print(f"P-value: {p_value:.4e}")

def p_to_stars(p):
    if p < 0.001:
        return "***"
    if p < 0.01:
        return "**"
    if p < 0.05:
        return "*"
    return "n.s."

print(f"Significance: {p_to_stars(p_value)}")

if p_value < 0.05:
    print("结论: P < 0.05，各特首之间的关键词使用密度存在显著的统计学差异。\n")
    
    # ==========================================
    # 4. 事后检验：Tukey HSD 
    # ==========================================
    print("--- Tukey HSD 事后检验 (谁和谁有显著差异？) ---")
    # alpha=0.05 代表 95% 置信区间
    tukey = pairwise_tukeyhsd(endog=df_stats['Density'], groups=df_stats['CE'], alpha=0.05)
    tukey_df = pd.DataFrame(data=tukey.summary().data[1:], columns=tukey.summary().data[0])
    tukey_df['p-adj'] = tukey_df['p-adj'].astype(float)
    tukey_df['sig'] = tukey_df['p-adj'].apply(p_to_stars)
    tukey_df = tukey_df.rename(columns={
        'group1': 'Group 1',
        'group2': 'Group 2',
        'meandiff': 'Mean Diff',
        'p-adj': 'p-adj',
        'lower': 'CI Low',
        'upper': 'CI High',
        'reject': 'Reject H0',
    })
    print(tukey_df.to_string(index=False))
else:
    print("结论: P >= 0.05，无法证明各特首之间的关键词密度存在显著差异。\n")

# 留存演讲数据，供后续合并可视化使用
df_raw_speeches = df_raw.copy()
df_stats_speeches = df_stats.copy()
```

```python
import scipy.stats as stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd
from matplotlib.patches import Patch

# ==========================================
# 1. 数据准备：计算单篇文档的Loyalty Signalling密度（仅 press releases）
# ==========================================
ce_name_map = {
    "Donald_Tsang": "Donald Tsang",
    "Leung_Chun-ying": "CY Leung",
    "Carrie_Lam": "Carrie Lam",
    "John_Lee": "John Lee",
    "Tung_Chee-hwa": "Tung Chee-hwa",
}

records = []
for fname, content in globals().get("john_lee_press_txt", {}).items():
    records.append({
        "CE": "John Lee",
        "Hits": sum(content.count(kw) for kw in loyalty_signaling_keywords),
        "Chars": len(content),
    })
for fname, content in globals().get("tung_chee_hwa_press_txt", {}).items():
    records.append({
        "CE": "Tung Chee-hwa",
        "Hits": sum(content.count(kw) for kw in loyalty_signaling_keywords),
        "Chars": len(content),
    })
for ce_folder, texts in globals().get("other_ce_press_txt", {}).items():
    ce_label = ce_name_map.get(ce_folder, ce_folder)
    for fname, content in texts.items():
        records.append({
            "CE": ce_label,
            "Hits": sum(content.count(kw) for kw in loyalty_signaling_keywords),
            "Chars": len(content),
        })

df_raw = pd.DataFrame(records)
if df_raw.empty:
    raise ValueError("No press-release documents loaded. Please run the press releases data loading cells first.")

# 在进行统计检验前，我们必须基于“每一篇演讲稿”计算密度，而不是年度总和
df_raw['Density'] = (df_raw['Hits'] / df_raw['Chars']) * 10000

# 剔除可能存在的缺失值或零字数异常文档
df_stats = df_raw[df_raw['Chars'] > 0].dropna(subset=['Density', 'CE'])

# ==========================================
# 2. 描述性统计与变异系数 (CV) 表格
# ==========================================
print("--- 历任特首用词频率统计汇总表 ---")
summary_stats = df_stats.groupby('CE')['Density'].agg(
    Count='count',
    Mean='mean',
    Median='median',
    Std='std'
).reset_index()

# 计算变异系数 CV (标准差 / 均值)
summary_stats['CV'] = summary_stats['Std'] / summary_stats['Mean']

# 按照时间顺序排序（假设你希望按照就任顺序查看）
# 这里我们自定义一个排序字典
ce_order = {"Tung Chee-hwa": 1, "Donald Tsang": 2, "CY Leung": 3, "Carrie Lam": 4, "John Lee": 5}
summary_stats['Order'] = summary_stats['CE'].map(ce_order)
summary_stats = summary_stats.sort_values('Order').drop('Order', axis=1).round(2)

print(summary_stats.to_string(index=False))
print("\n")

# ==========================================
# 3. 差异分析：单因素方差分析 (ANOVA)
# ==========================================
# 提取各特首的密度数据组
groups = [group['Density'].values for name, group in df_stats.groupby('CE')]

f_stat, p_value = stats.f_oneway(*groups)
print(f"--- 单因素方差分析 (ANOVA) ---")
print(f"F-statistic: {f_stat:.4f}")
print(f"P-value: {p_value:.4e}")

def p_to_stars(p):
    if p < 0.001:
        return "***"
    if p < 0.01:
        return "**"
    if p < 0.05:
        return "*"
    return "n.s."

print(f"Significance: {p_to_stars(p_value)}")

if p_value < 0.05:
    print("结论: P < 0.05，各特首之间的关键词使用密度存在显著的统计学差异。\n")
    
    # ==========================================
    # 4. 事后检验：Tukey HSD 
    # ==========================================
    print("--- Tukey HSD 事后检验 (谁和谁有显著差异？) ---")
    # alpha=0.05 代表 95% 置信区间
    tukey = pairwise_tukeyhsd(endog=df_stats['Density'], groups=df_stats['CE'], alpha=0.05)
    tukey_df = pd.DataFrame(data=tukey.summary().data[1:], columns=tukey.summary().data[0])
    tukey_df['p-adj'] = tukey_df['p-adj'].astype(float)
    tukey_df['sig'] = tukey_df['p-adj'].apply(p_to_stars)
    tukey_df = tukey_df.rename(columns={
        'group1': 'Group 1',
        'group2': 'Group 2',
        'meandiff': 'Mean Diff',
        'p-adj': 'p-adj',
        'lower': 'CI Low',
        'upper': 'CI High',
        'reject': 'Reject H0',
    })
    print(tukey_df.to_string(index=False))
else:
    print("结论: P >= 0.05，无法证明各特首之间的关键词密度存在显著差异。\n")

# 留存新闻稿数据，供合并可视化使用
df_raw_press = df_raw.copy()
df_stats_press = df_stats.copy()

# ==========================================
# 4. 按特首比较：press vs speeches 显著性检验
# ==========================================
if 'df_stats_speeches' not in globals() or df_stats_speeches.empty:
    raise ValueError("请先运行 CE speeches 的统计单元格，再运行本单元格以合并作图。")

comparison_rows = []
for ce_name in ce_order.keys():
    speech_vals = df_stats_speeches[df_stats_speeches['CE'] == ce_name]['Density'].dropna()
    press_vals = df_stats_press[df_stats_press['CE'] == ce_name]['Density'].dropna()
    if speech_vals.empty or press_vals.empty:
        continue
    t_stat, p_val = stats.ttest_ind(speech_vals, press_vals, equal_var=False)
    comparison_rows.append({
        "CE": ce_name,
        "Speeches N": len(speech_vals),
        "Press N": len(press_vals),
        "Speeches Mean": speech_vals.mean(),
        "Press Mean": press_vals.mean(),
        "Mean Diff": speech_vals.mean() - press_vals.mean(),
        "t-stat": t_stat,
        "p-value": p_val,
        "sig": p_to_stars(p_val),
    })

comparison_df = pd.DataFrame(comparison_rows)
if comparison_df.empty:
    print("No CE had both speeches and press releases for comparison.")
else:
    comparison_df = comparison_df.sort_values('CE')
    comparison_df[['Speeches Mean', 'Press Mean', 'Mean Diff', 't-stat']] = comparison_df[[
        'Speeches Mean', 'Press Mean', 'Mean Diff', 't-stat'
    ]].round(3)
    comparison_df['p-value'] = comparison_df['p-value'].apply(lambda v: float(f"{v:.4e}"))
    print("--- Press vs Speeches (per CE, Welch t-test) ---")
    print(comparison_df.to_string(index=False))

# ==========================================
# 5. 合并可视化：箱型图 + 散点 + 置信区间（深浅蓝）
# ==========================================
if 'df_raw_speeches' not in globals() or df_raw_speeches.empty:
    raise ValueError("请先运行 CE speeches 的统计单元格，再运行本单元格以合并作图。")
if 'df_raw_press' not in globals() or df_raw_press.empty:
    raise ValueError("请先运行 Press releases 的统计单元格，再运行本单元格以合并作图。")

df_stats_speeches = df_raw_speeches[df_raw_speeches['Chars'] > 0].dropna(subset=['Density', 'CE'])
df_stats_press = df_raw_press[df_raw_press['Chars'] > 0].dropna(subset=['Density', 'CE'])
print(f"Speeches docs: {len(df_stats_speeches)} | Press docs: {len(df_stats_press)}")
print(f"Speeches mean: {df_stats_speeches['Density'].mean():.3f} | Press mean: {df_stats_press['Density'].mean():.3f}")

order = ["Tung Chee-hwa", "Donald Tsang", "CY Leung", "Carrie Lam", "John Lee"]
combined = pd.concat(
    [
        df_stats_speeches.assign(Source="CE Speeches"),
        df_stats_press.assign(Source="Press Releases"),
    ],
    ignore_index=True,
)

sns.set_theme(style="whitegrid", context="talk", rc={"axes.edgecolor": "0.8", "grid.color": "0.9"})
fig, ax = plt.subplots(figsize=(12.8, 7.2))

source_palette = {
    "CE Speeches": "#2F5D8A",
    "Press Releases": "#B9CDE5",
}

# 箱型图：展示中位数、四分位数
sns.boxplot(
    x='CE', y='Density', data=combined,
    order=order,
    hue='Source',
    palette=source_palette,
    width=0.6,
    showfliers=False,
    dodge=True,
    boxprops={"alpha": 0.75, "linewidth": 1.0, "edgecolor": "#444444"},
    medianprops={"color": "black", "linewidth": 1.4},
    whiskerprops={"linewidth": 1.0, "color": "#444444"},
    capprops={"linewidth": 1.0, "color": "#444444"},
    ax=ax
 )

# 叠加散点图：每一个点代表一篇文本
sns.stripplot(
    x='CE', y='Density', data=combined,
    order=order,
    hue='Source',
    palette=source_palette,
    dodge=True,
    alpha=0.35,
    size=3.0,
    jitter=0.22,
    ax=ax
 )
if ax.legend_:
    ax.legend_.remove()

# 叠加均值与 95% 置信区间（bootstrap）
sns.pointplot(
    x='CE', y='Density', data=combined,
    order=order,
    hue='Source',
    estimator=np.mean,
    errorbar=("ci", 95),
    join=False,
    palette=source_palette,
    markers="D",
    scale=0.85,
    capsize=0.12,
    errwidth=1.0,
    dodge=0.4,
    ax=ax
 )
if ax.legend_:
    ax.legend_.remove()

# 仅保留必要的水平线（淡色）
ax.yaxis.grid(True, linestyle='--', linewidth=0.8, color='#D0D0D0', alpha=0.6)
ax.xaxis.grid(False)

# 文字采用 sentence case
ax.set_title("Distribution of Loyalty Keyword Density by Source and Chief Executive", fontsize=16, fontweight="bold", pad=18)
ax.set_xlabel("Chief Executive", fontsize=12, labelpad=10)
ax.set_ylabel("Loyalty Keyword Hits per 10k Characters", fontsize=12, labelpad=6)
sns.despine(ax=ax, top=True, right=True)

legend_handles = [
    Patch(facecolor=source_palette["CE Speeches"], edgecolor="#444444", label="CE speeches"),
    Patch(facecolor=source_palette["Press Releases"], edgecolor="#444444", label="Press releases"),
 ]
ax.legend(
    handles=legend_handles,
    loc="upper center",
    ncol=2,
    frameon=True,
    framealpha=0.92,
    edgecolor="lightgray",
    fontsize=11,
    title="Source",
    title_fontsize=10,
    bbox_to_anchor=(0.5, 1.02)
)

# 图注/脚注
fig.text(
    0.01, -0.08,
    "Note. Diamonds show mean with 95% CI (bootstrap, 1,000 resamples).",
    ha="left", va="bottom", fontsize=10
 )
fig.text(
    0.01, -0.12,
    "Significance: *** p<0.001, ** p<0.01, * p<0.05, n.s. p>=0.05.",
    ha="left", va="bottom", fontsize=10
 )

plt.tight_layout()
save_and_show(fig, "speeches_press_boxplot")
```

```python
import os
import re
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# ==================================================
# Year-by-year loyalty trend by Chief Executive
# ==================================================
# This cell rebuilds the six loyalty categories from the keyword cell in this notebook,
# counts keyword hits in each press-release text, and plots year-by-year densities.

notebook_candidates = []
if "project_path" in globals():
    notebook_candidates.append(Path(project_path) / "Code" / "Analysis_combined.ipynb")
notebook_candidates.append(Path(r"g:\My Drive\Study\HKU\POLI3148 Data Science in Politics and Public Administration\Final Project\Code\Analysis_combined.ipynb"))
notebook_candidates.append(Path.cwd() / "Analysis_combined.ipynb")
notebook_candidates.append(Path.cwd() / "Code" / "Analysis_combined.ipynb")

notebook_path = next((p for p in notebook_candidates if p.exists()), None)

# Prefer the canonical CE names in the chart.
ce_to_folder = {
    "Tung Chee-hwa": "Tung_Chee-hwa",
    "Donald Tsang": "Donald_Tsang",
    "Leung Chun-ying": "Leung_Chun-ying",
    "Carrie Lam": "Carrie_Lam",
    "John Lee": "John_Lee",
}

ce_label_map = {
    "Tung_Chee-hwa": "Tung Chee-hwa",
    "Donald_Tsang": "Donald Tsang",
    "Leung_Chun-ying": "Leung Chun-ying",
    "Carrie_Lam": "Carrie Lam",
    "John_Lee": "John Lee",
}

# Try to reuse the notebook's existing keyword cell by parsing the code source from disk.
def load_loyalty_categories_from_notebook(nb_path):
    with open(nb_path, "r", encoding="utf-8") as f:
        nb = json.load(f)

    target_source = None
    for cell in nb.get("cells", []):
        source = "".join(cell.get("source", []))
        if "personal_loyalty" in source and "Rhetorical_loyalty" in source:
            target_source = cell.get("source", [])
            break

    if target_source is None:
        raise ValueError("Could not find the loyalty keyword cell in the notebook.")

    category_aliases = {
        "personal_loyalty": "Personal loyalty",
        "Institutional_loyalty": "Institutional loyalty",
        "policy_loyalty": "Policy loyalty",
        "nationalistic_loyalty": "Nationalistic loyalty",
        "Ideological_loyalty": "Ideological loyalty",
        "Rhetorical_loyalty": "Rhetorical loyalty",
    }

    categories = {label: [] for label in category_aliases.values()}
    current = None
    for raw_line in target_source:
        line = raw_line.strip()
        if line.startswith('#"') and line.endswith('":'):
            key = line[2:-2]
            current = category_aliases.get(key)
            continue
        if current is None:
            continue
        matches = re.findall(r'"([^"]+)"|\'([^\']+)\'', line)
        for match in matches:
            token = match[0] or match[1]
            token = token.strip()
            if token:
                categories[current].append(token)

    # Remove duplicates while preserving order.
    for key, values in categories.items():
        seen = set()
        unique_values = []
        for value in values:
            if value not in seen:
                seen.add(value)
                unique_values.append(value)
        categories[key] = unique_values

    return categories


loyalty_categories = None
if notebook_path is not None:
    loyalty_categories = load_loyalty_categories_from_notebook(notebook_path)

# Fallback to the current kernel's keyword set if parsing fails or notebook is missing.
if not loyalty_categories or not any(loyalty_categories.values()):
    if "loyalty_signaling_keywords" in globals() and isinstance(loyalty_signaling_keywords, (set, list, tuple)):
        fallback_keywords = list(loyalty_signaling_keywords)
        loyalty_categories = {"Overall loyalty terms": fallback_keywords}
    else:
        raise ValueError("No loyalty signaling keywords available.")

# Find the raw press-release directories.
base_press_release = Path(project_path) / "Data" / "Data_Raw" / "Speeches_and_Articles"
if not base_press_release.exists():
    base_press_release = Path(r"g:\My Drive\Study\HKU\POLI3148 Data Science in Politics and Public Administration\Final Project\Data\Data_Raw\Speeches_and_Articles")

ce_folders = [folder for folder in ce_to_folder.values() if (base_press_release / folder).exists()]


def extract_year_from_filename(file_name):
    name = str(file_name)
    patterns = [
        r'(19\d{2}|20\d{2})[-_](\d{2})[-_](\d{2})',
        r'(\d{2})[-_](\d{2})[-_](19\d{2}|20\d{2})',
        r'(19\d{2}|20\d{2})',
    ]
    for pattern in patterns:
        match = re.search(pattern, name)
        if match:
            groups = match.groups()
            if len(groups) == 3:
                if groups[0].isdigit() and len(groups[0]) == 4:
                    return int(groups[0])
                if groups[2].isdigit() and len(groups[2]) == 4:
                    return int(groups[2])
            if len(groups) == 1:
                return int(groups[0])
    return None


def count_keyword_hits(text, keywords):
    total = 0
    for kw in keywords:
        if not kw:
            continue
        total += len(re.findall(re.escape(kw), text))
    return total

records = []
for folder_name in ce_folders:
    folder_path = base_press_release / folder_name
    for lang_name in ["Traditional_Chinese", "Simplified_Chinese"]:
        lang_path = folder_path / lang_name
        if not lang_path.exists():
            continue
        for file_path in lang_path.glob("*.txt"):
            try:
                content = file_path.read_text(encoding="utf-8", errors="ignore")
            except Exception:
                continue
            year = extract_year_from_filename(file_path.name)
            if year is None:
                header_match = re.search(r"^(?:Date|日期)[:：]\s*(.+)$", content, flags=re.MULTILINE)
                if header_match:
                    year_match = re.search(r'(19\d{2}|20\d{2})', header_match.group(1))
                    if year_match:
                        year = int(year_match.group(1))
            if year is None:
                continue
            ce_label = ce_label_map.get(folder_name, folder_name)
            char_count = len(content)
            for category_name, keyword_list in loyalty_categories.items():
                hits = count_keyword_hits(content, keyword_list)
                density = (hits / char_count * 10000) if char_count else 0
                records.append({
                    "CE": ce_label,
                    "Year": year,
                    "Category": category_name,
                    "Hits": hits,
                    "Chars": char_count,
                    "Density": density,
                    "Language": lang_name,
                })

trend_df = pd.DataFrame(records)
if trend_df.empty:
    raise ValueError("No press-release records were collected for the loyalty trend chart.")

# Combine the two language copies by averaging the densities across the same year/CE/category.
trend_df = (
    trend_df.groupby(["CE", "Year", "Category"], as_index=False)
    .agg({"Hits": "sum", "Chars": "sum", "Density": "mean"})
)

category_order = [
    "Personal loyalty",
    "Institutional loyalty",
    "Policy loyalty",
    "Nationalistic loyalty",
    "Ideological loyalty",
    "Rhetorical loyalty",
]
existing_categories = [c for c in category_order if c in trend_df["Category"].unique()]
if not existing_categories:
    existing_categories = sorted(trend_df["Category"].unique())

ce_order = [ce for ce in ["Tung Chee-hwa", "Donald Tsang", "Leung Chun-ying", "Carrie Lam", "John Lee"] if ce in trend_df["CE"].unique()]

sns.set_theme(style="whitegrid", rc={"axes.edgecolor": "0.8", "grid.color": "0.9"})
# Use sharey=True to unify the y-axis across all subplots
fig, axes = plt.subplots(2, 3, figsize=(22, 12), sharex=True, sharey=True)
axes = axes.flatten()

# A more aesthetic palette
palette = {
    "Tung Chee-hwa": "#3498DB",   # blue
    "Donald Tsang": "#2ECC71",    # green
    "Leung Chun-ying": "#F1C40F", # yellow 
    "Carrie Lam": "#E67E22",      # orange
    "John Lee": "#E74C3C",        # red
}

for idx, category_name in enumerate(existing_categories):
    ax = axes[idx]
    subset = trend_df[trend_df["Category"] == category_name].copy()
    
    for ce_name in ce_order:
        ce_data = subset[subset["CE"] == ce_name].sort_values("Year")
        if ce_data.empty:
            continue
            
        color = palette.get(ce_name, "#333333")
        
        # Plot lines with a slight shadow/glow effect
        ax.plot(
            ce_data["Year"],
            ce_data["Density"],
            marker="o",
            linewidth=2.5,
            markersize=6,
            markeredgecolor="white",
            markeredgewidth=1.5,
            label=ce_name,
            color=color,
            alpha=0.9
        )
        
        # Optionally fill under the line to make it pop slightly
        ax.fill_between(
            ce_data["Year"], 
            0, 
            ce_data["Density"], 
            color=color, 
            alpha=0.05
        )
        
    ax.set_title(category_name, fontsize=15, fontweight="bold", pad=12)
    ax.set_xlabel("Year", fontsize=12)
    ax.set_ylabel("Density (Hits / 10k chars)", fontsize=12)
    
    all_years = sorted(trend_df["Year"].unique())
    # Display every 2 years to avoid crowding
    ax.set_xticks(np.arange(min(all_years), max(all_years)+1, 2))
    ax.tick_params(axis='x', rotation=45, labelsize=11)
    ax.tick_params(axis='y', labelsize=11)
    ax.grid(True, linestyle='--', alpha=0.6)
    
    # Remove top and right spines
    sns.despine(ax=ax, top=True, right=True)

# Remove any unused axes
for j in range(len(existing_categories), len(axes)):
    fig.delaxes(axes[j])

# Common legend
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles, labels, 
    loc="upper center", 
    ncol=min(len(ce_order), 5), 
    frameon=True, 
    framealpha=0.9,
    edgecolor="lightgray",
    fontsize=13,
    bbox_to_anchor=(0.5, 0.94)
)

fig.suptitle("Evolution of Loyalty Signalling Density in CE Speeches by Category", fontsize=22, fontweight="bold", y=0.99)
fig.tight_layout(rect=[0, 0, 1, 0.9])
save_and_show(fig, "speeches_by_category")

print("✅ 已生成优化后的 year-by-year loyalty 对比图。")
```

```python
import os
import re
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# ==================================================
# Year-by-year loyalty trend by Chief Executive
# ==================================================
# This cell rebuilds the six loyalty categories from the keyword cell in this notebook,
# counts keyword hits in each press-release text, and plots year-by-year densities.

notebook_candidates = []
if "project_path" in globals():
    notebook_candidates.append(Path(project_path) / "Code" / "Press_Release_analysis.ipynb")
notebook_candidates.append(Path(r"g:\My Drive\Study\HKU\POLI3148 Data Science in Politics and Public Administration\Final Project\Code\Press_Release_analysis.ipynb"))
notebook_candidates.append(Path.cwd() / "Press_Release_analysis.ipynb")
notebook_candidates.append(Path.cwd() / "Code" / "Press_Release_analysis.ipynb")

notebook_path = next((p for p in notebook_candidates if p.exists()), None)
if notebook_path is None:
    raise FileNotFoundError("Could not locate Press_Release_analysis.ipynb for keyword parsing.")

# Prefer the canonical CE names in the chart.
ce_to_folder = {
    "Tung Chee-hwa": "Tung_Chee-hwa",
    "Donald Tsang": "Donald_Tsang",
    "Leung Chun-ying": "Leung_Chun-ying",
    "Carrie Lam": "Carrie_Lam",
    "John Lee": "John_Lee",
}

ce_label_map = {
    "Tung_Chee-hwa": "Tung Chee-hwa",
    "Donald_Tsang": "Donald Tsang",
    "Leung_Chun-ying": "Leung Chun-ying",
    "Carrie_Lam": "Carrie Lam",
    "John_Lee": "John Lee",
}

# Try to reuse the notebook's existing keyword cell by parsing the code source from disk.
def load_loyalty_categories_from_notebook(nb_path):
    with open(nb_path, "r", encoding="utf-8") as f:
        nb = json.load(f)

    target_source = None
    for cell in nb.get("cells", []):
        source = "".join(cell.get("source", []))
        if "personal_loyalty" in source and "Rhetorical_loyalty" in source:
            target_source = cell.get("source", [])
            break

    if target_source is None:
        raise ValueError("Could not find the loyalty keyword cell in the notebook.")

    category_aliases = {
        "personal_loyalty": "Personal loyalty",
        "Institutional_loyalty": "Institutional loyalty",
        "policy_loyalty": "Policy loyalty",
        "nationalistic_loyalty": "Nationalistic loyalty",
        "Ideological_loyalty": "Ideological loyalty",
        "Rhetorical_loyalty": "Rhetorical loyalty",
    }

    categories = {label: [] for label in category_aliases.values()}
    current = None
    for raw_line in target_source:
        line = raw_line.strip()
        if line.startswith('#"') and line.endswith('":'):
            key = line[2:-2]
            current = category_aliases.get(key)
            continue
        if current is None:
            continue
        matches = re.findall(r'"([^"]+)"|\'([^\']+)\'', line)
        for match in matches:
            token = match[0] or match[1]
            token = token.strip()
            if token:
                categories[current].append(token)

    # Remove duplicates while preserving order.
    for key, values in categories.items():
        seen = set()
        unique_values = []
        for value in values:
            if value not in seen:
                seen.add(value)
                unique_values.append(value)
        categories[key] = unique_values

    return categories


loyalty_categories = load_loyalty_categories_from_notebook(notebook_path)

# Fallback to the current kernel's keyword set if parsing fails for any reason.
if not loyalty_categories or not any(loyalty_categories.values()):
    if "loyalty_signaling_keywords" in globals() and isinstance(loyalty_signaling_keywords, (set, list, tuple)):
        fallback_keywords = list(loyalty_signaling_keywords)
        loyalty_categories = {"Overall loyalty terms": fallback_keywords}
    else:
        raise ValueError("No loyalty signaling keywords available.")

# Find the raw press-release directories.
base_press_release = Path(project_path) / "Data" / "Data_Raw" / "Press_Release"
if not base_press_release.exists():
    base_press_release = Path(r"g:\My Drive\Study\HKU\POLI3148 Data Science in Politics and Public Administration\Final Project\Data\Data_Raw\Press_Release")

ce_folders = [folder for folder in ce_to_folder.values() if (base_press_release / folder).exists()]


def extract_year_from_filename(file_name):
    name = str(file_name)
    patterns = [
        r'(19\d{2}|20\d{2})[-_](\d{2})[-_](\d{2})',
        r'(\d{2})[-_](\d{2})[-_](19\d{2}|20\d{2})',
        r'(19\d{2}|20\d{2})',
    ]
    for pattern in patterns:
        match = re.search(pattern, name)
        if match:
            groups = match.groups()
            if len(groups) == 3:
                if groups[0].isdigit() and len(groups[0]) == 4:
                    return int(groups[0])
                if groups[2].isdigit() and len(groups[2]) == 4:
                    return int(groups[2])
            if len(groups) == 1:
                return int(groups[0])
    return None


def count_keyword_hits(text, keywords):
    total = 0
    for kw in keywords:
        if not kw:
            continue
        total += len(re.findall(re.escape(kw), text))
    return total

records = []
for folder_name in ce_folders:
    folder_path = base_press_release / folder_name
    for lang_name in ["Traditional_Chinese", "Simplified_Chinese"]:
        lang_path = folder_path / lang_name
        if not lang_path.exists():
            continue
        for file_path in lang_path.glob("*.txt"):
            try:
                content = file_path.read_text(encoding="utf-8", errors="ignore")
            except Exception:
                continue
            year = extract_year_from_filename(file_path.name)
            if year is None:
                header_match = re.search(r"^(?:Date|日期)[:：]\s*(.+)$", content, flags=re.MULTILINE)
                if header_match:
                    year_match = re.search(r'(19\d{2}|20\d{2})', header_match.group(1))
                    if year_match:
                        year = int(year_match.group(1))
            if year is None:
                continue
            ce_label = ce_label_map.get(folder_name, folder_name)
            char_count = len(content)
            for category_name, keyword_list in loyalty_categories.items():
                hits = count_keyword_hits(content, keyword_list)
                density = (hits / char_count * 10000) if char_count else 0
                records.append({
                    "CE": ce_label,
                    "Year": year,
                    "Category": category_name,
                    "Hits": hits,
                    "Chars": char_count,
                    "Density": density,
                    "Language": lang_name,
                })

trend_df = pd.DataFrame(records)
if trend_df.empty:
    raise ValueError("No press-release records were collected for the loyalty trend chart.")

# Combine the two language copies by averaging the densities across the same year/CE/category.
trend_df = (
    trend_df.groupby(["CE", "Year", "Category"], as_index=False)
    .agg({"Hits": "sum", "Chars": "sum", "Density": "mean"})
)

category_order = [
    "Personal loyalty",
    "Institutional loyalty",
    "Policy loyalty",
    "Nationalistic loyalty",
    "Ideological loyalty",
    "Rhetorical loyalty",
]
existing_categories = [c for c in category_order if c in trend_df["Category"].unique()]
if not existing_categories:
    existing_categories = sorted(trend_df["Category"].unique())

ce_order = [ce for ce in ["Tung Chee-hwa", "Donald Tsang", "Leung Chun-ying", "Carrie Lam", "John Lee"] if ce in trend_df["CE"].unique()]

sns.set_theme(style="whitegrid", rc={"axes.edgecolor": "0.8", "grid.color": "0.9"})
# Use sharey=True to unify the y-axis across all subplots
fig, axes = plt.subplots(2, 3, figsize=(22, 12), sharex=True, sharey=True)
axes = axes.flatten()

# A more aesthetic palette
palette = {
    "Tung Chee-hwa": "#3498DB",   # blue
    "Donald Tsang": "#2ECC71",    # green
    "Leung Chun-ying": "#F1C40F", # yellow 
    "Carrie Lam": "#E67E22",      # orange
    "John Lee": "#E74C3C",        # red
}

for idx, category_name in enumerate(existing_categories):
    ax = axes[idx]
    subset = trend_df[trend_df["Category"] == category_name].copy()
    
    for ce_name in ce_order:
        ce_data = subset[subset["CE"] == ce_name].sort_values("Year")
        if ce_data.empty:
            continue
            
        color = palette.get(ce_name, "#333333")
        
        # Plot lines with a slight shadow/glow effect
        ax.plot(
            ce_data["Year"],
            ce_data["Density"],
            marker="o",
            linewidth=2.5,
            markersize=6,
            markeredgecolor="white",
            markeredgewidth=1.5,
            label=ce_name,
            color=color,
            alpha=0.9
        )
        
        # Optionally fill under the line to make it pop slightly
        ax.fill_between(
            ce_data["Year"], 
            0, 
            ce_data["Density"], 
            color=color, 
            alpha=0.05
        )
        
    ax.set_title(category_name, fontsize=15, fontweight="bold", pad=12)
    ax.set_xlabel("Year", fontsize=12)
    ax.set_ylabel("Density (Hits / 10k chars)", fontsize=12)
    
    all_years = sorted(trend_df["Year"].unique())
    # Display every 2 years to avoid crowding
    ax.set_xticks(np.arange(min(all_years), max(all_years)+1, 2))
    ax.tick_params(axis='x', rotation=45, labelsize=11)
    ax.tick_params(axis='y', labelsize=11)
    ax.grid(True, linestyle='--', alpha=0.6)
    
    # Remove top and right spines
    sns.despine(ax=ax, top=True, right=True)

# Remove any unused axes
for j in range(len(existing_categories), len(axes)):
    fig.delaxes(axes[j])

# Common legend
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles, labels, 
    loc="upper center", 
    ncol=min(len(ce_order), 5), 
    frameon=True, 
    framealpha=0.9,
    edgecolor="lightgray",
    fontsize=13,
    bbox_to_anchor=(0.5, 0.94)
)

fig.suptitle("Evolution of Loyalty Signalling Density in Press Releases by Category", fontsize=22, fontweight="bold", y=0.99)
fig.tight_layout(rect=[0, 0, 1, 0.9])
save_and_show(fig, "press_by_category")

print("✅ 已生成优化后的 year-by-year loyalty 对比图。")
```
